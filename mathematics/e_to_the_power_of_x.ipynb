```python
import matplotlib.pyplot as plt
import numpy as np

# Okabe-Ito Color-Blind Friendly Palette
okabe_ito = {
    'black': '#000000',
    'orange': '#E69F00',
    'sky_blue': '#56B4E9',
    'bluish_green': '#009E73',
    'yellow': '#F0E442',
    'blue': '#0072B2',
    'vermillion': '#D55E00',
    'reddish_purple': '#CC79A7',
}

# Define range
x = np.linspace(-2, 2.5, 500)
y_exp = np.exp(x)
y_deriv = np.exp(x)  # d/dx(e^x) = e^x

plt.figure(figsize=(9, 6))

# Plot e^x
plt.plot(
    x,
    y_exp,
    label=r'$f(x) = e^x$',
    color=okabe_ito['blue'],
    linewidth=2.5,
)

# Plot derivative d/dx(e^x)
plt.plot(
    x,
    y_deriv,
    label=r'$\frac{d}{dx}(e^x) = e^x$',
    color=okabe_ito['orange'],
    linestyle='--',
    linewidth=2.5,
)

# Highlight e at x = 1
e_val = np.e
plt.axhline(
    y=e_val,
    color=okabe_ito['black'],
    linestyle=':',
    alpha=0.6,
    label=f'$e \\approx {e_val:.5f}$',
)
plt.axvline(x=1, color=okabe_ito['black'], linestyle=':', alpha=0.6)

# Scatter plot at (1, e) and (0, 1)
plt.scatter([1], [e_val], color=okabe_ito['vermillion'], zorder=5)
plt.annotate(
    f'$(1, e) \\approx (1, {e_val:.3f})$',
    xy=(1, e_val),
    xytext=(0.2, e_val + 0.5),
    arrowprops=dict(arrowstyle='->', lw=1.5, color=okabe_ito['vermillion']),
    fontsize=11,
    fontweight='bold',
)

plt.scatter([0], [1], color=okabe_ito['reddish_purple'], zorder=5)
plt.annotate(
    '$(0, 1)$',
    xy=(0, 1),
    xytext=(-1.2, 1.5),
    arrowprops=dict(arrowstyle='->', lw=1.5, color=okabe_ito['reddish_purple']),
    fontsize=11,
)

plt.title(
    r'Visualization of $e$, $e^x$, and $\frac{d}{dx}(e^x)$',
    fontsize=14,
    pad=12,
)
plt.xlabel('$x$', fontsize=12)
plt.ylabel('$y$', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11, loc='upper left')
plt.xlim(-2, 2.5)
plt.ylim(-0.5, 12)

plt.tight_layout()
plt.show()  # Renders directly in Jupyter Notebook output cell
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

x = np.linspace(-2, 2.5, 500)

def plot_tangent_secant(a=0.0, h=1.0, show_tangent=True, show_secant=True):
    b = a + h
    fa, fb = np.exp(a), np.exp(b)

    fig, ax = plt.subplots(figsize=(9, 6))
    ax.plot(x, np.exp(x), color=okabe_ito['blue'], linewidth=2.5, label=r'$f(x) = e^x$')

    if show_tangent:
        # Tangent at x=a: slope = f'(a) = e^a
        ax.plot(x, fa * (x - a) + fa, color=okabe_ito['orange'], linestyle='--',
                linewidth=2, label=rf'Tangent at $x={a:.2f}$ (slope $={fa:.3f}$)')

    if show_secant and abs(h) > 1e-9:
        # Secant through (a, e^a) and (a+h, e^(a+h))
        slope = (fb - fa) / h
        ax.plot(x, slope * (x - a) + fa, color=okabe_ito['bluish_green'], linewidth=2,
                label=rf'Secant, $h={h:.2f}$ (slope $={slope:.3f}$)')
        ax.scatter([b], [fb], color=okabe_ito['reddish_purple'], zorder=5, s=60)

    ax.scatter([a], [fa], color=okabe_ito['vermillion'], zorder=5, s=60)

    ax.set_xlim(-2, 2.5)
    ax.set_ylim(-0.5, 12)
    ax.set_xlabel('$x$', fontsize=12)
    ax.set_ylabel('$y$', fontsize=12)
    ax.set_title(r'Tangent and Secant Lines of $e^x$', fontsize=14, pad=12)
    ax.grid(True, linestyle='--', alpha=0.5)
    ax.legend(fontsize=10, loc='upper left')
    plt.show()

interact(
    plot_tangent_secant,
    a=FloatSlider(value=0.0, min=-2, max=2, step=0.05, description='a (point)'),
    h=FloatSlider(value=1.0, min=-2, max=2, step=0.05, description='h (gap)'),
);
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Reuses `okabe_ito` from cell 1

x_vals_deg = np.array([0, 30, 45, 60, 90])
x_vals_rad = np.radians(x_vals_deg)

x_continuous = np.linspace(0, np.pi / 2, 1000)
threshold = 10

# (name, function, color, has_asymptote)
# Yellow is skipped: it has poor contrast on a white background.
trig_funcs = [
    (r"$\sin(e^x)$", np.sin,                      okabe_ito['blue'],           False),
    (r"$\cos(e^x)$", np.cos,                      okabe_ito['bluish_green'],   False),
    (r"$\tan(e^x)$", np.tan,                      okabe_ito['reddish_purple'], True),
    (r"$\csc(e^x)$", lambda t: 1 / np.sin(t),     okabe_ito['orange'],         True),
    (r"$\sec(e^x)$", lambda t: 1 / np.cos(t),     okabe_ito['vermillion'],     True),
    (r"$\cot(e^x)$", lambda t: 1 / np.tan(t),     okabe_ito['sky_blue'],       True),
]

fig, axes = plt.subplots(2, 3, figsize=(15, 9))
fig.suptitle(
    r"Visualizing Trigonometric Functions of $e^x$ "
    r"(for $x \in [0, \pi/2]$ radians / $0^\circ$ to $90^\circ$)",
    fontsize=14,
    fontweight="bold",
)

e_cont = np.exp(x_continuous)
e_pts = np.exp(x_vals_rad)

for ax, (title, func, color, asymptote) in zip(axes.flat, trig_funcs):
    with np.errstate(divide="ignore", invalid="ignore"):
        data = func(e_cont)
        pt_y = func(e_pts)

    if asymptote:
        data = np.where(np.abs(data) > threshold, np.nan, data)

    ax.plot(x_continuous, data, color=color, linewidth=2.5)

    for x_r, y_val, deg in zip(x_vals_rad, pt_y, x_vals_deg):
        if not asymptote or abs(y_val) <= threshold:
            ax.plot(x_r, y_val, "o", color=okabe_ito['black'], markersize=6)
            ax.annotate(
                f"{deg}°", (x_r, y_val),
                textcoords="offset points", xytext=(0, 8),
                ha="center", fontsize=8,
            )

    ax.set_title(title, fontsize=12, fontweight="bold")
    ax.grid(True, linestyle="--", alpha=0.5)
    ax.axhline(0, color=okabe_ito['black'], linewidth=0.8, linestyle=":")
    ax.set_xlabel("x (radians)", fontsize=10)
    ax.set_ylim((-10, 10) if asymptote else (-1.2, 1.2))

plt.tight_layout()
plt.savefig("trig_e_x_plot.png", dpi=300)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(9, 6))

# Slope field: at every point (x, y), the slope is simply y
gx, gy = np.meshgrid(np.linspace(-2, 2.5, 30), np.linspace(-6, 12, 28))
dx = 0.09
ax.quiver(
    gx, gy, np.full_like(gx, dx), gy * dx,   # segment = (dx, slope * dx)
    angles='xy', scale_units='xy', scale=1, pivot='mid',
    headwidth=0, headlength=0, headaxislength=0,  # draw plain segments, no arrowheads
    width=0.003, color=okabe_ito['orange'], alpha=0.8,
)

xx = np.linspace(-2, 2.5, 400)

# Other solutions y = C * e^x
for i, C in enumerate([-2, -1, -0.5, 0.5, 2]):
    ax.plot(
        xx, C * np.exp(xx),
        color=okabe_ito['sky_blue'], linewidth=1.8,
        label=r'$y = Ce^x$ (other $C$)' if i == 0 else None,
    )

# The solution through (0, 1) is e^x itself
ax.plot(xx, np.exp(xx), color=okabe_ito['blue'], linewidth=3,
        label=r'$y = e^x$  ($C = 1$)')

# C = 0 is the flat solution
ax.axhline(0, color=okabe_ito['black'], linestyle=':', alpha=0.6)

# Slope equals height: at x = 1, both are e
ax.scatter([1], [np.e], color=okabe_ito['vermillion'], zorder=5, s=60)
ax.annotate(
    r'height $= e$, slope $= e$',
    xy=(1, np.e), xytext=(-1.6, 7.5),
    arrowprops=dict(arrowstyle='->', lw=1.5, color=okabe_ito['vermillion']),
    fontsize=11, fontweight='bold',
)

ax.set_xlim(-2, 2.5)
ax.set_ylim(-6, 12)
ax.set_xlabel('$x$', fontsize=12)
ax.set_ylabel('$y$', fontsize=12)
ax.set_title(r"Slope Field of $y' = y$: Growth Proportional to Size", fontsize=14, pad=12)
ax.grid(True, linestyle='--', alpha=0.4)
ax.legend(fontsize=11, loc='upper left')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

x = np.linspace(-2, 2, 400)

# (label, base, color)
bases = [
    (r'$2^x$',  2.0,    okabe_ito['orange']),
    (r'$e^x$',  np.e,   okabe_ito['blue']),
    (r'$3^x$',  3.0,    okabe_ito['bluish_green']),
]

# Left panel: curves and their tangent lines at x = 0
for label, b, color in bases:
    slope = np.log(b)  # d/dx b^x at x = 0 is ln(b)
    ax1.plot(x, b**x, color=color, linewidth=2.5,
             label=rf'{label}  (slope at 0 $= {slope:.3f}$)')
    ax1.plot(x, 1 + slope * x, color=color, linestyle='--', linewidth=1.5, alpha=0.8)

ax1.scatter([0], [1], color=okabe_ito['vermillion'], zorder=5, s=70)
ax1.annotate('$(0, 1)$: every $b^x$ passes here', xy=(0, 1), xytext=(-1.9, 5.5),
             arrowprops=dict(arrowstyle='->', lw=1.5, color=okabe_ito['vermillion']),
             fontsize=11)
ax1.set_xlim(-2, 2)
ax1.set_ylim(-1, 8)
ax1.set_xlabel('$x$', fontsize=12)
ax1.set_ylabel('$y$', fontsize=12)
ax1.set_title('Tangent at $x = 0$ for different bases (dashed)', fontsize=13, pad=10)
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(fontsize=10, loc='upper left')

# Right panel: slope at x = 0 as a function of the base
b_range = np.linspace(1.01, 5, 400)
ax2.plot(b_range, np.log(b_range), color=okabe_ito['reddish_purple'], linewidth=2.5,
         label=r'slope at $x=0$ $= \ln(b)$')
ax2.axhline(1, color=okabe_ito['black'], linestyle=':', alpha=0.6)
ax2.axvline(np.e, color=okabe_ito['black'], linestyle=':', alpha=0.6)

for label, b, color in bases:
    ax2.scatter([b], [np.log(b)], color=color, s=90, zorder=5, edgecolor=okabe_ito['black'])

ax2.annotate(r'$b = e$: slope is exactly $1$', xy=(np.e, 1), xytext=(3.2, 0.35),
             arrowprops=dict(arrowstyle='->', lw=1.5, color=okabe_ito['vermillion']),
             fontsize=11, fontweight='bold')
ax2.set_xlim(1, 5)
ax2.set_ylim(0, 1.8)
ax2.set_xlabel('base $b$', fontsize=12)
ax2.set_ylabel(r'slope of $b^x$ at $x = 0$', fontsize=12)
ax2.set_title(r'$e$ is the base whose slope at $0$ equals $1$', fontsize=13, pad=10)
ax2.grid(True, linestyle='--', alpha=0.5)
ax2.legend(fontsize=10, loc='upper left')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

x = np.linspace(-4, 4, 600)

def taylor_exp(x, N):
    """Partial sum of the series: sum of x^n / n! for n = 0..N."""
    total = np.zeros_like(x)
    term = np.ones_like(x)
    for n in range(N + 1):
        total += term
        term = term * x / (n + 1)
    return total

def plot_taylor(N=3):
    approx = taylor_exp(x, N)
    err = np.maximum(np.abs(np.exp(x) - approx), 1e-17)  # floor so the log axis stays finite

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

    # Left: function vs polynomial approximation
    ax1.plot(x, np.exp(x), color=okabe_ito['blue'], linewidth=3, label=r'$e^x$')
    ax1.plot(x, approx, color=okabe_ito['orange'], linestyle='--', linewidth=2.5,
             label=rf'Taylor, degree $N={N}$')
    ax1.set_xlim(-4, 4)
    ax1.set_ylim(-5, 50)
    ax1.set_xlabel('$x$', fontsize=12)
    ax1.set_ylabel('$y$', fontsize=12)
    ax1.set_title(r'$e^x = \sum_{n=0}^{\infty} \frac{x^n}{n!}$, truncated after $N$ terms',
                  fontsize=13, pad=10)
    ax1.grid(True, linestyle='--', alpha=0.5)
    ax1.legend(fontsize=11, loc='upper left')

    # Right: absolute error on a log axis
    ax2.semilogy(x, err, color=okabe_ito['vermillion'], linewidth=2.5)
    ax2.set_xlim(-4, 4)
    ax2.set_ylim(1e-16, 1e3)
    ax2.set_xlabel('$x$', fontsize=12)
    ax2.set_ylabel(r'$|e^x - T_N(x)|$', fontsize=12)
    ax2.set_title('Approximation error (log scale)', fontsize=13, pad=10)
    ax2.grid(True, which='both', linestyle='--', alpha=0.4)

    plt.tight_layout()
    plt.show()

interact(plot_taylor, N=IntSlider(value=3, min=0, max=15, step=1, description='N (degree)'));
```

```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

def plot_euler(theta=np.pi / 4):
    c, s = np.cos(theta), np.sin(theta)
    t = np.linspace(0, 2 * np.pi, 400)

    fig, (ax1, ax2) = plt.subplots(
        1, 2, figsize=(14, 6), gridspec_kw={'width_ratios': [1, 1.4]}
    )

    # Left: e^{i*theta} on the unit circle
    ax1.plot(np.cos(t), np.sin(t), color=okabe_ito['black'], alpha=0.35, linewidth=1.5)
    arc = np.linspace(0, theta, 200)
    ax1.plot(np.cos(arc), np.sin(arc), color=okabe_ito['blue'], linewidth=4)
    ax1.plot([0, c], [0, s], color=okabe_ito['blue'], linewidth=2)
    ax1.plot([0, c], [0, 0], color=okabe_ito['bluish_green'], linewidth=4,
             label=r'$\cos\theta$ (real part)')
    ax1.plot([c, c], [0, s], color=okabe_ito['vermillion'], linewidth=4,
             label=r'$\sin\theta$ (imaginary part)')
    ax1.scatter([c], [s], color=okabe_ito['blue'], s=90, zorder=5, edgecolor=okabe_ito['black'])
    ax1.axhline(0, color=okabe_ito['black'], linewidth=0.8)
    ax1.axvline(0, color=okabe_ito['black'], linewidth=0.8)
    ax1.set_xlim(-1.4, 1.4)
    ax1.set_ylim(-1.4, 1.4)
    ax1.set_aspect('equal')
    ax1.set_xlabel('Real', fontsize=12)
    ax1.set_ylabel('Imaginary', fontsize=12)
    ax1.set_title(rf'$e^{{i\theta}} = {c:.3f} + {s:.3f}i$', fontsize=13, pad=10)
    ax1.grid(True, linestyle='--', alpha=0.5)
    ax1.legend(fontsize=10, loc='lower left')

    # Right: the same projections unrolled over theta
    ax2.plot(t, np.cos(t), color=okabe_ito['bluish_green'], linewidth=2.5, label=r'$\cos\theta$')
    ax2.plot(t, np.sin(t), color=okabe_ito['vermillion'], linewidth=2.5, label=r'$\sin\theta$')
    ax2.axvline(theta, color=okabe_ito['black'], linestyle=':', alpha=0.7)
    ax2.scatter([theta], [c], color=okabe_ito['bluish_green'], s=80, zorder=5, edgecolor=okabe_ito['black'])
    ax2.scatter([theta], [s], color=okabe_ito['vermillion'], s=80, zorder=5, edgecolor=okabe_ito['black'])
    ax2.set_xticks([0, np.pi / 2, np.pi, 3 * np.pi / 2, 2 * np.pi])
    ax2.set_xticklabels(['0', r'$\pi/2$', r'$\pi$', r'$3\pi/2$', r'$2\pi$'])
    ax2.set_xlim(0, 2 * np.pi)
    ax2.set_ylim(-1.3, 1.3)
    ax2.set_xlabel(r'$\theta$ (radians)', fontsize=12)
    ax2.set_title(r"Euler's formula: $e^{i\theta} = \cos\theta + i\sin\theta$", fontsize=13, pad=10)
    ax2.grid(True, linestyle='--', alpha=0.5)
    ax2.legend(fontsize=11, loc='upper right')

    plt.tight_layout()
    plt.show()

interact(
    plot_euler,
    theta=FloatSlider(value=np.pi / 4, min=0, max=2 * np.pi, step=0.05, description=r'θ (rad)'),
);
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-2, 3, 500)

# (k, color)
rates = [
    (-1.0, okabe_ito['reddish_purple']),
    (0.5,  okabe_ito['sky_blue']),
    (1.0,  okabe_ito['blue']),
    (2.0,  okabe_ito['vermillion']),
]

def make_label(k):
    kstr = {1.0: '', -1.0: '-'}.get(k, f'{k:g}')
    base = rf'$e^{{{kstr}x}}$'
    t = np.log(2) / abs(k)
    verb = 'doubles' if k > 0 else 'halves'
    return f'{base}  ({verb} every {t:.2f})'

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))

for k, color in rates:
    y = np.exp(k * x)
    ax1.plot(x, y, color=color, linewidth=2.5, label=make_label(k))
    ax2.plot(x, y, color=color, linewidth=2.5, label=make_label(k))

# Linear scale: curves
ax1.set_ylim(0, 20)
ax1.set_xlim(-2, 3)
ax1.set_xlabel('$x$', fontsize=12)
ax1.set_ylabel('$y$ (linear scale)', fontsize=12)
ax1.set_title(r'Linear scale: $e^{kx}$ curves up or down', fontsize=13, pad=10)
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(fontsize=10, loc='upper left')

# Log scale: the same functions become straight lines
ax2.set_yscale('log')
ax2.set_ylim(1e-2, 1e3)
ax2.set_xlim(-2, 3)
ax2.axhline(1, color=okabe_ito['black'], linestyle=':', alpha=0.6)
ax2.set_xlabel('$x$', fontsize=12)
ax2.set_ylabel('$y$ (log scale)', fontsize=12)
ax2.set_title(r'Log scale: each line has slope $k$ (steeper = faster)', fontsize=13, pad=10)
ax2.grid(True, which='both', linestyle='--', alpha=0.4)
ax2.legend(fontsize=10, loc='upper left')

plt.tight_layout()
plt.show()
```
